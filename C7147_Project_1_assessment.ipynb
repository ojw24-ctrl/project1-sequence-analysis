## 🧬 Load reference and reads data into variables
Below, we provide some helper code to load the reference from `reference.txt` and the reads from `reads.csv`.

❗ **Do not change the function’s behavior** (you may read it, of course!).

The code creates two variables from the loaded files:
- **reference**: a string containing the 200-nt reference sequence
- **reads**: a list of dictionaries containing the 6,000 reads

````python
##### Constants

REF_PATH = "reference.txt"
READS_PATH = "reads.csv"

##### Functions

def load_reference(path: str) -> str:
    """Load reference nucleotide sequence from file at path to reference string."""
    with open(path) as f:
        lines = [ln.strip() for ln in f if ln.strip()]
    if lines[0].startswith('>'):
        lines = lines[1:]
    ref = "".join(lines).upper()
    if len(ref) != 200:
        raise ValueError(f"Reference must be 200 nt, got {len(ref)}")
    return ref


def load_reads(path: str) -> list:
    """Load sample NGS reads from file at path to reads list."""
    with open(path, newline="") as f:
        reader = csv.DictReader(f)
        reads = list(reader)
    return reads

##### Main code

# Load reference and reads
reference = load_reference(REF_PATH)
reads = load_reads(READS_PATH)

# Report data structures
print("Reference length:", len(reference))
print("Total no. of reads:", len(reads))
print("Keys in each read dictionary:", reads[0].keys())
print("Samples loaded:", sorted({read["sample"] for read in reads}))

# Print sample sequences
print("\nSample read data:")
for row in reads[:5]:
    print(row["read_id"], row["sample"], row['group'], row["sequence"])
print("...")
for row in reads[-5:]:
    print(row["read_id"], row["sample"], row['group'], row["sequence"])
````

## Step 1 — Validate each read and record any errors
## Function: `validate_read(sequence, reference)`

### 👍 You can assume:
- `reference` is a non-empty string containing only uppercase A/C/G/T.
- `sequence` is a string, but nothing about its length or contents.

### 🎯 Aims:
- Check that an individual read is the same length as the reference and contains only valid characters (A/C/G/T)
- "Repair" any reads that contain valid lower-case characters -> upper-case
- Return a dictionary with the following keys:

    - `sequence`: the upper-case version of the sequence input, including when invalid.
    - `valid`: `True` only when the length matches the reference and every character is A/C/G/T.
    - `errors`: a list of strings. Include `"wrong_length"` first if lengths differ, then `"invalid_characters"` if any character is outside A/C/G/T. Include both when both apply. A valid read has `[]`.

### 🔎 Expected outputs

Examples for reference `"ATGC"`:
| Input (read) | Input type | Output |
|---|---|:---|
| `"ATGC"` | Correct | `{"sequence": "ATGC", "valid": True, "errors": []}` |
| `"atgc"` | Lowercase | `{"sequence": "ATGC", "valid": True, "errors": []}` |
| `""` | Empty read | `{"sequence": "", "valid": False, "errors": ["wrong_length"]}` |
| `"ATNC"` | Invalid character | `{"sequence": "ATNC", "valid": False, "errors": ["invalid_characters"]}` |
| `"ATN"` | Multiple errors | `{"sequence": "ATN", "valid": False, "errors": ["wrong_length", "invalid_characters"]}` |


Invalid reads should not crash your program.

### 💻 Enter your code here

````python
def validate_read(sequence, reference):
    """validate_read checks whether a sequence is suitable for comparison with the reference (same length and valid bases containing ACTG).
    It will also "repair" any sequence with lower case bases in the sequence so that they are upper case.
    Args:
     sequence <- string, can be any length
     reference <- non-empty string, containing ONLY the four bases A, C, G, or T, serves as a point of comparison to sequence."""
    # Write your code here.
    sequence = sequence.upper()
    errors = []
    error_count = 0
    #this part is to check whether the length of the sequence is the same as the reference.
    if len(sequence) != len(reference):
      errors.append("wrong_length")
      error_count += 1
    #this part is to check the different bases and whether they are valid (A, T, C, or G)
    valid_base = set("ATCG")
    for base in sequence:
      if base not in valid_base:
        errors.append("invalid_characters")
        error_count += 1
        break

    read_validation = {"sequence": sequence, "valid": error_count == 0, "errors": errors}

    return read_validation






````

## Step 2 — Find every mismatch in a valid read
## Function: `find_mutations(sequence, reference)`

### 👍 You can assume:
- Inputs `sequence` and `reference` are valid (i.e. both are strings of the same length and contain only upper-case bases, i.e. A/C/G/T)

### 🎯 Aims:
- Return a list containing one dictionary per mismatch, in ascending position order.
- Each dictionary has exactly these keys:
    - `position`: position of the mutation (integer), numbered from 1.
    - `reference_base`: one-character string from reference at that position.
    - `read_base`: one-character string from the read at that position.

- Return `[]` if the read matches the reference.

### 🔎 Expected outputs

Example: `find_mutations("GTGA", "ATGC")` returns:
```python
[{"position": 1, "reference_base": "A", "read_base": "G"},
 {"position": 4, "reference_base": "C", "read_base": "A"}]
```


### 💻 Enter your code here

````python
def find_mutations(sequence, reference):
    """find_mutations finds mutations in comparison to the reference sequence.
     It will also find the position of the mutation and what base it has mutated to."""
    # Write your code here.
    mutation = []
    for i in range(len(sequence)):
      if sequence[i] != reference[i]:
        mutation.append({"position": i + 1,
                       "reference_base": reference[i],
                       "read_base": sequence[i]})

    return mutation


````

## Step 3 — Analyse all reads
## Function: `analyse_reads(reads, reference)`

### 👍 You can assume:

- Each read dictionary should contain the following keys (values below are examples):
```python
{
    "read_id": "read_0001",
    "sample": "sample1",
    "group": "treatment",
    "sequence": "GTGA",             # normalised upper-case sequence
    "status": "valid",              # "valid" or "invalid"
    "errors": [],                   # validation error codes
    "mutations": [
        {"position": 1, "reference_base": "A", "read_base": "G"},
        {"position": 4, "reference_base": "C", "read_base": "A"}
    ]

```

### 🔎 Expected outputs
- Valid reads have `status="valid"` and `mutations` listing all mutations in the dictionary format shown above.
- Invalid reads have `status="invalid"`, the relevant error codes, and `mutations=[]`.
- Empty input returns `[]`.

### 💻 Enter your code here

````python
def analyse_reads(reads, reference):
    """analyse_reads will provide a dictionary listing:
    - your sequence's validity
    - mutation
    - mutation location.
    Invalid sequences will not compare the sequence with the reference for mutations, but will list why it is invalid under errors.
    Args:
     reads <- dictionary list, the reads that you would like to compare (with keys: read_id, sample, group, and sequence)
     reference <- string, the sequence that will serve as a reference in comparison to reads"""
    # Write your code here.
    reading = []
    for read in reads:
      validate = validate_read(read["sequence"], reference)
      if validate["valid"]:
        mutations = find_mutations(validate["sequence"], reference)
        status = "valid"
      else:
        mutations = []
        status = "invalid"
      reading.append({"read_id": read["read_id"],
                   "sample": read["sample"],
                   "group": read["group"],
                   "sequence": validate["sequence"],
                     "status": status,
                     "errors": validate["errors"],
                     "mutations": mutations})
    return reading

````

### 💻 Enter your code here

````python

# Write your code here.

def get_top_positions(position_counts):
  """get_top_positions is a helper function that will provide the top three positions with the most mismatches"""
  return sorted(position_counts, key=lambda p: (-position_counts[p], p))[:3]

analysed_reads = analyse_reads(reads, reference)

def summarise_samples(analysed_reads):
  """summarise_samples will return a dictionary containing information about reads that have been analysed by the analyse_reads function.
  Args:
   analysed_reads -> dictionary, analysed samples from both the given sequence, and a reference sequence """
  sum_samples = {}

  for read in analysed_reads:
    sample_name = read["sample"]
    group = read["group"]
    if sample_name not in sum_samples:
      sum_samples[sample_name] = {"group": group,
                                 "valid_reads": 0,
                                 "invalid_reads": 0,
                                 "mismatch_reads": 0,
                                 "mismatch_percentage": None,
                                 "mismatch_events": 0,
                                 "substitution_counts":{},
                                 "position_counts": {},
                                 "top_positions": []}

  #counts valid readings
    if read["status"] == "valid":
      sum_samples[sample_name]["valid_reads"] += 1
    else:
      sum_samples[sample_name]["invalid_reads"] += 1
      continue
  #counting mismatch events from mutations
    if len(read["mutations"]) > 0:
      sum_samples[sample_name]["mismatch_reads"] += 1
  #counting the substitutions and mismatch reads
    for m in read["mutations"]:
      position = m["position"]
      sum_samples[sample_name]["mismatch_events"] += 1
      substitution_counts = m["reference_base"] + ">" + m["read_base"]
      if substitution_counts in sum_samples[sample_name]["substitution_counts"]:
        sum_samples[sample_name]["substitution_counts"][substitution_counts] += 1
      else:
        sum_samples[sample_name]["substitution_counts"][substitution_counts] = 1
   #counting positions
      pos_count = sum_samples[sample_name]["position_counts"]
      if position in pos_count:
        pos_count[position] += 1
      else:
        pos_count[position] = 1

 #calculates the mismatch percentage for only valid samples, and top positions.

  for sample_name in sum_samples:
    if sum_samples[sample_name]["valid_reads"] > 0:
      sum_samples[sample_name]["mismatch_percentage"] = (sum_samples[sample_name]["mismatch_reads"] /
                                                         sum_samples[sample_name]["valid_reads"])*100

    pos_count = sum_samples[sample_name]["position_counts"]
    sum_samples[sample_name]["top_positions"] = get_top_positions(sum_samples[sample_name]["position_counts"])


  return sum_samples


````

## Function:

- Select the **top 3 positions with the highest combined counts** (or fewer if there are less than three mismatches), choosing the earlier reference position in the case of a tie.

### 🧮 Sample-level calculations

### 📊 Statistical comparison

For each position, pass the sample percentages to the supplied function as a list (one per group) and store the result, e.g.:

```python
statistical_result = calculate_statistics(control_percentages, treated_percentages)
```

Each list must contain **one percentage per biological sample**.

The supplied function returns:

```python
{"statistic": float, "p_value": float}
```

It returns `None` values when the comparison cannot be calculated.

````python
#@title 📊 calculate_statistics function

from scipy import stats
import math
import warnings


def calculate_statistics(control_percentages, treated_percentages):
    """Compare independent biological samples using a two-sided Welch's t-test."""
    unavailable = {"statistic": None, "p_value": None}

    # Require at least two samples per group and finite numerical values.
    if len(control_percentages) < 2 or len(treated_percentages) < 2:
        return unavailable

    values = list(control_percentages) + list(treated_percentages)
    if any(value is None or not math.isfinite(value) for value in values):
        return unavailable

    # No within-group variation: the t-test cannot estimate uncertainty.
    if (
        len(set(control_percentages)) == 1
        and len(set(treated_percentages)) == 1
    ):
        return unavailable

    with warnings.catch_warnings():
        warnings.simplefilter("ignore", RuntimeWarning)
        result = stats.ttest_ind(
            treated_percentages,
            control_percentages,
            equal_var=False,
            alternative="two-sided"
        )

    if not (
        math.isfinite(result.statistic)
        and math.isfinite(result.pvalue)
    ):
        return unavailable

    return {
        "statistic": float(result.statistic),
        "p_value": float(result.pvalue)
    }
````

Return a list with one dictionary per selected position:

    }

Means are sample mismatch percentages; `difference` is **treated mean − control mean**, in percentage points. Use `None` for results that cannot be calculated.

Print **one brief line per position**, reporting group means, their difference, sample numbers and p-value.

### 💻 Enter your code here

````python
# Write your code here.

#making a helper function that will add the mismatch counts at every position
def mismatch_counts(sample_summary, groups):
  """The combined_counts function is a helper function primarily used to create the compare_positions function.
  It adds together the mismatches of each position per group (gathered from position_count in sample_summary)."""
  com_count = {}
  for name in sample_summary:
    sample_name = sample_summary[name]
    if sample_name["group"] in groups:
      for pos in sample_name["position_counts"]:
        if pos in com_count:
          com_count[pos] += sample_name["position_counts"][pos]
        else:
          com_count[pos] = sample_name["position_counts"][pos]
  return com_count


def compare_positions(sample_summary, control_group="control", treated_group="treatment"):
  """compare_positions returns a dictionary per selected position, containing the n, mean, difference, statistic, and p-value.
  The statistics and p-value values are calculated by a two-sided welch test, generated by the calculate_statistics function.
  Args:
   sample_summary -> the dictionary returned by the summarise_samples function, for further information on this, go to the doctring for the function summarise_samples
   control_group, treated_group -> strings, respectively the two groups you would like a comparison between, with the name of the group specified - for example control_group="control"

  Note:
  If there are no values for certain keys like control_mean, treated_mean and difference, it will output None."""
  groups = [control_group, treated_group]
  com_count = mismatch_counts(sample_summary, groups)
  top_positions = get_top_positions(com_count)
  pos_comp = []
#percentage calculations
  for pos in top_positions:
    percentage_control = []
    percentage_treated = []
    for name in sample_summary:
      sample_name = sample_summary[name]
      if sample_name["valid_reads"] == 0:
        continue
      if pos in sample_name["position_counts"]:
        count = sample_name["position_counts"][pos]
      else:
        count = 0
      percentages = count/sample_name["valid_reads"]*100
      if sample_name["group"] == control_group:
        percentage_control.append(percentages)
      elif sample_name["group"] == treated_group:
        percentage_treated.append(percentages)
#means
    control_mean = None
    treated_mean = None
    difference = None
    if len(percentage_control) > 0:
      control_mean = sum(percentage_control)/len(percentage_control)
    if len(percentage_treated) > 0:
      treated_mean = sum(percentage_treated)/len(percentage_treated)
#difference
    if treated_mean is not None and control_mean is not None:
      difference = treated_mean - control_mean
#making the dictionary from all the info gathered before
    statistical_result = calculate_statistics(percentage_control, percentage_treated)
    pos_comp.append({"position": pos,
              "control_n": len(percentage_control),
              "treated_n": len(percentage_treated),
              "control_mean": control_mean,
              "treated_mean": treated_mean,
              "difference": difference,
              "statistic": statistical_result["statistic"],
              "p_value": statistical_result["p_value"]})

  return pos_comp






````
